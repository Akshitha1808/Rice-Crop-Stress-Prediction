```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Variables used to construct the stress index
stress_variables = [
    "NDVI",
    "EVI",
    "Average_Soil_Moisture",
    "Rainfall",
    "Average_LST",
    "Maximum_Temperature",
    "Evapotranspiration"
]

# Min-Max normalization
scaler = MinMaxScaler()

scaled = scaler.fit_transform(df[stress_variables])

scaled_df = pd.DataFrame(
    scaled,
    columns=stress_variables,
    index=df.index
)

# Convert variables into stress-oriented scores
# Higher score = greater stress

df["NDVI_Stress"] = 1 - scaled_df["NDVI"]

df["EVI_Stress"] = 1 - scaled_df["EVI"]

df["Soil_Moisture_Stress"] = (
    1 - scaled_df["Average_Soil_Moisture"]
)

df["Rainfall_Stress"] = (
    1 - scaled_df["Rainfall"]
)

df["LST_Stress"] = scaled_df["Average_LST"]

df["Temperature_Stress"] = (
    scaled_df["Maximum_Temperature"]
)

df["ET_Stress"] = scaled_df["Evapotranspiration"]

# Crop Stress Index
df["Crop_Stress_Index"] = (
    df["NDVI_Stress"] +
    df["EVI_Stress"] +
    df["Soil_Moisture_Stress"] +
    df["Rainfall_Stress"] +
    df["LST_Stress"] +
    df["Temperature_Stress"] +
    df["ET_Stress"]
) / 7

print(df["Crop_Stress_Index"].describe())
```

```python
# ==========================================================
# Create 30–60 Day Early Prediction Dataset
# ==========================================================

# Variables that will be used as predictors
predictor_cols = [
    "NDVI",
    "EVI",
    "Day_LST",
    "Night_LST",
    "Soil_Moisture_AM",
    "Soil_Moisture_PM",
    "Rainfall",
    "Mean_Temperature",
    "Maximum_Temperature",
    "Minimum_Temperature",
    "Relative_Humidity",
    "Wind_Speed",
    "Evapotranspiration",
    "Average_LST",
    "Temperature_Range",
    "Average_Soil_Moisture",
    "NDVI_EVI_Ratio",
    "Water_Availability",
    "Soil_Moisture_Difference"
]


# ----------------------------------------------------------
# Kharif
# September = ~60 days before November
# October   = ~30 days before November
# ----------------------------------------------------------

kharif_oct = df[
    (df["Crop_Season"] == "Kharif") &
    (df["Month"] == 10)
].copy()

kharif_sep = df[
    (df["Crop_Season"] == "Kharif") &
    (df["Month"] == 9)
].copy()

kharif_nov = df[
    (df["Crop_Season"] == "Kharif") &
    (df["Month"] == 11)
].copy()


# Rename October variables as 30-day predictors
kharif_oct = kharif_oct[
    ["District", "Year"] + predictor_cols
].copy()

kharif_oct = kharif_oct.rename(
    columns={
        col: col + "_Lag30"
        for col in predictor_cols
    }
)


# Rename September variables as 60-day predictors
kharif_sep = kharif_sep[
    ["District", "Year"] + predictor_cols
].copy()

kharif_sep = kharif_sep.rename(
    columns={
        col: col + "_Lag60"
        for col in predictor_cols
    }
)


# Harvest target = November
kharif_target = kharif_nov[
    ["District", "Year", "Harvest_Stress"]
].copy()


# Merge Kharif
kharif_model = kharif_oct.merge(
    kharif_sep,
    on=["District", "Year"],
    how="inner"
)

kharif_model = kharif_model.merge(
    kharif_target,
    on=["District", "Year"],
    how="inner"
)

kharif_model["Crop_Season"] = "Kharif"


# ----------------------------------------------------------
# Rabi/Summer
# February = ~60 days before April
# March    = ~30 days before April
# ----------------------------------------------------------

rabi_mar = df[
    (df["Crop_Season"] == "Rabi_Summer") &
    (df["Month"] == 3)
].copy()

rabi_feb = df[
    (df["Crop_Season"] == "Rabi_Summer") &
    (df["Month"] == 2)
].copy()

rabi_apr = df[
    (df["Crop_Season"] == "Rabi_Summer") &
    (df["Month"] == 4)
].copy()


# Rename March variables as 30-day predictors
rabi_mar = rabi_mar[
    ["District", "Year"] + predictor_cols
].copy()

rabi_mar = rabi_mar.rename(
    columns={
        col: col + "_Lag30"
        for col in predictor_cols
    }
)


# Rename February variables as 60-day predictors
rabi_feb = rabi_feb[
    ["District", "Year"] + predictor_cols
].copy()

rabi_feb = rabi_feb.rename(
    columns={
        col: col + "_Lag60"
        for col in predictor_cols
    }
)


# Harvest target = April
rabi_target = rabi_apr[
    ["District", "Year", "Harvest_Stress"]
].copy()


# Merge Rabi/Summer
rabi_model = rabi_mar.merge(
    rabi_feb,
    on=["District", "Year"],
    how="inner"
)

rabi_model = rabi_model.merge(
    rabi_target,
    on=["District", "Year"],
    how="inner"
)

rabi_model["Crop_Season"] = "Rabi_Summer"


# ----------------------------------------------------------
# Combine both seasons
# ----------------------------------------------------------

model_df = pd.concat(
    [kharif_model, rabi_model],
    ignore_index=True
)


print("Model dataset shape:", model_df.shape)

print("\nSeason distribution:")
print(model_df["Crop_Season"].value_counts())

print("\nTarget distribution:")
print(model_df["Harvest_Stress"].value_counts())

print("\nMissing values:")
print(model_df.isnull().sum().sum())
```

```python
# Training = 2018–2023
train_mask = model_df["Year"] <= 2023

# Validation = 2024
val_2024_mask = model_df["Year"] == 2024

# Create training and 2024 datasets
X_train_full = model_df.loc[
    train_mask,
    feature_names
].copy()

X_2024 = model_df.loc[
    val_2024_mask,
    feature_names
].copy()

y_train_full = model_df.loc[
    train_mask,
    "Harvest_Stress"
].copy()

print("Training shape:", X_train_full.shape)
print("2024 shape:", X_2024.shape)
```

```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# --------------------------------------------------
# 1. Load shapefile
# --------------------------------------------------

shapefile_path = "/content/drive/MyDrive/Rice_Crop_Stress/Cleaned_Data/Shapefile-AP/gadm41_IND_2.shp"

ap_districts = gpd.read_file(shapefile_path)

# Keep only Andhra Pradesh
ap_districts = ap_districts[
    ap_districts["NAME_1"] == "Andhra Pradesh"
].copy()

# --------------------------------------------------
# 2. Keep only selected study districts
# --------------------------------------------------

study_districts = [
    "West Godavari",
    "East Godavari",
    "Krishna",
    "Guntur",
    "Srikakulam",
    "Vizianagaram",
    "Nellore",
    "Prakasam",
    "Chittoor",
    "Visakhapatnam"
]

ap_districts = ap_districts[
    ap_districts["NAME_2"].isin(study_districts)
].copy()

# --------------------------------------------------
# 3. IMPORTANT: Dissolve multiple polygons
#    into one geometry per district
# --------------------------------------------------

ap_districts = ap_districts.dissolve(
    by="NAME_2",
    as_index=False
)

print("Number of districts:", len(ap_districts))
print(ap_districts["NAME_2"].tolist())

# --------------------------------------------------
# 4. Select year and season
# --------------------------------------------------

year = 2025
season = "Kharif"

map_data = df[
    (df["Year"] == year) &
    (df["Crop_Season"] == season) &
    (df["District"].isin(study_districts))
].copy()

map_data["District"] = map_data["District"].str.strip()

# --------------------------------------------------
# 5. Check district-level data
# --------------------------------------------------

print("\nData used for mapping:")
print(
    map_data[
        ["District", "Year", "Crop_Season", "Harvest_Stress"]
    ].to_string(index=False)
)

# --------------------------------------------------
# 6. If multiple rows exist per district,
#    keep one stress value per district
# --------------------------------------------------

map_data = (
    map_data[
        ["District", "Harvest_Stress"]
    ]
    .drop_duplicates(subset="District")
)

# --------------------------------------------------
# 7. Merge with spatial boundaries
# --------------------------------------------------

merged = ap_districts.merge(
    map_data,
    left_on="NAME_2",
    right_on="District",
    how="left"
)

print("\nFinal mapping data:")
print(
    merged[
        ["NAME_2", "Harvest_Stress"]
    ].to_string(index=False)
)

# --------------------------------------------------
# 8. Define colors
# --------------------------------------------------

stress_colors = {
    "Normal": "#8BC34A",
    "Moderate Stress": "#FFC107",
    "Severe Stress": "#E53935"
}

# --------------------------------------------------
# 9. Create map
# --------------------------------------------------

fig, ax = plt.subplots(
    figsize=(10, 10)
)

# Plot each stress class
for stress_class, color in stress_colors.items():

    subset = merged[
        merged["Harvest_Stress"] == stress_class
    ]

    if not subset.empty:
        subset.plot(
            ax=ax,
            color=color,
            edgecolor="black",
            linewidth=0.8
        )

# Plot missing-data districts
missing = merged[
    merged["Harvest_Stress"].isna()
]

if not missing.empty:
    missing.plot(
        ax=ax,
        color="lightgrey",
        edgecolor="black",
        linewidth=0.8
    )

# --------------------------------------------------
# 10. Add district names
# --------------------------------------------------

for _, row in merged.iterrows():

    if row.geometry is not None and not row.geometry.is_empty:

        point = row.geometry.representative_point()

        ax.annotate(
            row["NAME_2"],
            xy=(point.x, point.y),
            ha="center",
            va="center",
            fontsize=8,
            fontweight="bold"
        )

# --------------------------------------------------
# 11. Title
# --------------------------------------------------

ax.set_title(
    f"Spatial Distribution of Rice Crop Stress\n"
    f"Andhra Pradesh – {year} {season}",
    fontsize=16,
    fontweight="bold"
)

# --------------------------------------------------
# 12. Legend
# --------------------------------------------------

legend_elements = [
    Patch(
        facecolor="#8BC34A",
        edgecolor="black",
        label="Normal"
    ),
    Patch(
        facecolor="#FFC107",
        edgecolor="black",
        label="Moderate Stress"
    ),
    Patch(
        facecolor="#E53935",
        edgecolor="black",
        label="Severe Stress"
    )
]

if not missing.empty:
    legend_elements.append(
        Patch(
            facecolor="lightgrey",
            edgecolor="black",
            label="No Data"
        )
    )

ax.legend(
    handles=legend_elements,
    title="Crop Stress",
    loc="lower left"
)

ax.set_axis_off()

plt.tight_layout()

# --------------------------------------------------
# 13. Save
# --------------------------------------------------

output_path = (
    "/content/drive/MyDrive/Rice_Crop_Stress/"
    "Spatial_Crop_Stress_2025_Kharif.png"
)

plt.savefig(
    output_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nMap saved successfully:")
print(output_path)
```
